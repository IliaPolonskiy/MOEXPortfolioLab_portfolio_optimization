```python
import numpy as np
import matplotlib.pyplot as plt
from method_classes.classSharpFrontier import SharpFrontier

# 1. Генерируем тестовые данные (10 активов)
np.random.seed(42)
n = 10
mu = np.random.uniform(0.05, 0.25, n)
A = np.random.randn(n, n)
S = A @ A.T * 0.01 

# 2. Инициализируем наш новый класс
frontier = SharpFrontier(mu, S)

# 3. Находим оптимальный портфель для безрисковой ставки 3%
opt = frontier.get_analytical_max_sharpe(rf=0.03)
print(f"Sharpe: {opt['sharpe']:.4f}")
print(f"Доходность: {opt['returns']:.2%}")
print(f"Риск: {opt['volatilities']:.2%}")

# 4. Строим график
fig, ax = plt.subplots(figsize=(10, 6))
frontier.plot_efficient_frontier(ax=ax, rf=0.03)
plt.show()
```
